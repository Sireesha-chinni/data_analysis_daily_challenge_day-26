# Coding Challenge Set: Strings & Tuples

Challenge 1: Find Longest Student Name
    
    
Find the longest name from tuple.


Input
students = ("Anu", "Ravikumar", "Jo")

```python
students = ("Anu", "Ravikumar", "Jo")
lenth = []

for i in students:
    lenth.append(len(i))
    
    
print(students[lenth.index(max(lenth))])
```

Challenge 2: Join Subjects into Single String
    
    
Display subjects in UI


Input

subjects = ("Math", "Science", "English")

```python
subjects = ("Math", "Science", "English")
on_str=""

for i in subjects:
    on_str = on_str +i

print(on_str)
```

Challenge 3: Reverse Each Student Name
    
    
Data masking requirement


Input
students = ("Anu", "Ravi")

```python
students = ("Anu", "Ravi")
res = []

for i in students:
    res.append(i[::-1])
    
students=tuple(res)
print(students)
```

Challenge 4: Count Vowels in Each Name
    
    
Analyze student names


Input
students = ("Anu", "Ravi")

```python
students = ("Anu", "Ravi","Sireesha")
vowel = ("A","a","E","e","I","i","O","o","U","u")
count_vo =[]


for i in students:
    num = 0
    for j in i:
        if j in vowel:
            num= num+1
    count_vo.append(num)

print(count_vo)
```

Challenge 5: Convert Tuple to Uppercase String
    
    
Prepare report header


Input
courses = ("python", "sql", "power bi")

```python
courses = ("python", "sql", "power bi")
cour =[]

for i in courses:
    cour.append(i.upper())

courses = tuple(cour)
print(courses)
```
